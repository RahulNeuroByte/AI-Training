## TYPES OF MATRICES
## 1. ONES MATRIX ...having 1 as element
## 2  ZERO MATRIX
## 3. IDENTITY MATRIX....diagonal value has 1 and non diagonal value will be 0

```python
# create an array using np.ones() matrix
np.ones([5, 6])    # shape = (5 rows , 6 = columns)
```

```python
np.ones([5, 6], dtype = np.int8)
```

```python
np.zeros([5, 6])   # shape (5 rows & 6 columns)
```

```python
np.identity(7)       # square matrix with diagonal values as 1 & non diagonal values are 0s.
```

```python
np.linspace(0, 1, 11) #generate 11 evenly spaced values between 0 and 1 (inclusive of both 0 and 1)
```

```python
np.linspace(0, 1, 6)    # generate evenly spaced numbers over a specified interval.
```

```python
# np.nan = missing value
arr_8 = np.array([10, 20, 30, np.nan, 50, 60, np.nan, 80, 90])
```

```python
np.isnan(arr_8)    # np.nan() returns boolean value; True for missing value & false for non-missing
```

```python
np.where(np.isnan(arr_8))
```

```python
np.where(arr_8 == 60)   # np.where(condition) : returns index position where condition is True
```
