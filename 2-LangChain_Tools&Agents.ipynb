11. Agent with Multiple tools

```python
agent = initialize_agent(
    tools=[search_tool, youtube_search_tool, calculator_tool],
    llm=llm,
    agent = AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True,
    max_iterations=5,
    early_stopping_method="generate",
    handle_parsing_errors=True
)
```

```python
# Run the agent in loop with a query that requires multiple tools
```

```python
while True:
    query = input("Enter your query (or type 'exit' to quit): ")
    if query.lower() == 'exit':
        break
    response = agent.invoke(query)
    print(response)

# to come out from this loop enter: exit
```
