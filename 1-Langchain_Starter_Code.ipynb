```python
# Output Template
from langchain_core.output_parsers import StrOutputParser, JsonOutputParser
```

```python
system_template = "Classify sentence into Positive, Negative or Neutral."
```

```python
from langchain_core.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate.from_messages([
    ("system", system_template),
    ("user", "{input}")
])
```

```python
parser = StrOutputParser()
```

```python
#in Language This is chain- LECL (Langchain Expression Chain Language)
# It is a combination of prompt, llm and parser
chain = prompt | llm | parser
```

```python
result = chain.invoke({"input": "I purchased a new phone yesterday and it is amazing!"})
 
```

6. Tools
* Tools are external features or capabilities that an LLM/Agent can call to perform actions beyond generating text response.

```python
from langchain_core.tools import Tool
```

```python
def calculate(query):
    return str(eval(query))
```

```python
calculator_tool = Tool(
    name="Calculator",
    func=calculate,
    description="Useful for when you need to answer questions about math."
)
```

```python
calculator_tool.invoke("1845 * 2341 / 10 + 67543 % 2")
```

```python
from langchain_core.tools import tool
 
@tool
def calculator(query: str) -> str:
    """Useful for when you need to answer questions about math."""
    try:
        return str(eval(query))
    except Exception as e:
        return f"Invalid Mathematical Expression: {e}"
```

```python
# Bind Tools to LLM
llm_with_tools = llm.bind_tools([calculator])
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant that can perform calculations."),
    ("user", "{input}")
])
```

```python
parser = StrOutputParser()
chain = prompt | llm_with_tools | parser
```

```python
question = "What is the result of 1845 * 2341 / 10 + 67543 % 2?"
```

```python
response = chain.invoke({"input" : question})
```
