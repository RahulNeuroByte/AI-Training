```python
#create a checkerboard pattern of 8x8 using numpy and indexing concept
import numpy as np
checkerboard = np.zeros((8, 8),dtype=int)
checkerboard[1::2, ::2] = 1
checkerboard[::2, 1::2] = 1
print(checkerboard)
```

```python
#converting above checker board into color image using matplotlib
import matplotlib.pyplot as plt
plt.imshow(checkerboard, cmap='gray')   
```

## NUMPY MATHEMATICAL FUNCTIONS

```python
# apply numpy mathematical functions in a new 2d array 
arr_4 = np.arange(1, 51).reshape(5, 10)
print(arr_4)
```

```python
# applying min, max, mean, median, std, sum, prod, cumsum, cumprod functions on the array
print("Minimum value:", np.min(arr_4))
print("Maximum value:", np.max(arr_4))
print("Mean value:", np.mean(arr_4))
print("Median value:", np.median(arr_4))
print("Standard deviation:", np.std(arr_4))
print("Sum value:", np.sum(arr_4))
print("Product value:", np.prod(arr_4))
print("Cumulative sum:", np.cumsum(arr_4))
print("Cumulative product:", np.cumprod(arr_4))
```

```python
# print sum row wise and column wise
print(arr_4)
print("Sum of each row:", np.sum(arr_4, axis=1))
print("Sum of each column:", np.sum(arr_4, axis=0))
```

```python
diagonal=np.diag(arr_4)
print("Diagonal elements of the array:", diagonal)
np.sum(np.diag(arr_4))
```

```python
# Broadcasting in numpy
arr_5 = np.arange(1, 10).reshape(3, 3)
print(arr_5)
```

```python
arr_6 = np.array([1, 2, 3])
print(arr_5 + arr_6)  # Broadcasting addition
```

```python
array_2 = np.arange(1, 17).reshape(4, 4)
 
print(array_2)
print("\n")

array_3 = np.array([[1, 2, 3, 4]])
print(array_3)
print("\n")

print(array_3.shape)
print(array_2.shape)

print("Broadcasting :", array_2 + array_3)  # Broadcasting addition
 
```

```python
customer_final = customer_df.merge(cdr_df, on = 'customer_id', how = 'inner')
```

```python
customer_df.merge(cdr_df, on = 'customer_id', how = 'right')
```
