## Matplotlib
A low-level plotting library in Python used for creating fundamentals plots - Bar, Line, Pie, Histogram, etc.
Plots will be static and interactive
It provides control over each aspect of plots(axes, labels, colors, style, background-style)
## Seaborn
A high level statistical data visualization library build on top of Matplotlib
Provides a simpler and more elegant syntax for complex plots
Specialized in statistical plots - distribution plot, heatmap, pair-plots, boxplot, violin plots etc.

## 2. Read datasets

```python
auto_df = pd.read_csv("https://raw.githubusercontent.com/ammishra08/Machine-Learning/refs/heads/master/Datasets/Automobile_price_data__Raw_.csv",
                            header = 0, sep = ' ,')                         
```

```python
auto_df = auto_df.replace('?', np.nan)
```

```python
auto_df.info()
```

```python
auto_df.drop(['symboling','normalized-losses'], axis = 1, inplace = True)
```

```python
# chnage datatype of multiple columns using pd.to_numeric and apply(function_name) method
cols = ['bore', 'stroke', 'horsepower', 'peak-rpm', 'price']
auto_df[cols] = auto_df[cols].apply(pd.to_numeric, errors = 'coerce')
```
