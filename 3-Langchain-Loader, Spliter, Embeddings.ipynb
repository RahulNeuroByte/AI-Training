* ## LangChain
* IT provides different types of text splitters to break large documents into smaller chunks that fit within model context windows.Main Types of Text Splitters
* RecursiveCharacterTextSplitter: The default and recommended choice. 
* It splits text recursively using a list of separators (like paragraphs, sentences, and words) to keep related context together.
* CharacterTextSplitter: A basic splitter that breaks text based on a single user-defined character or separator (such as \n\n) and a fixed character length.
* TokenTextSplitter: Splits text based on exact token counts rather than characters, aligning chunks with how language models read data

```python
char_splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=50,
    separators=["\n\n", "\n", ".", ""])
```

```python
char_chunks = char_splitter.split_documents(documents)
```
